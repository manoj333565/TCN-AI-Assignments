```python
# Q1.
# @log_calls
# Write a decorator that prints the function name, its arguments and its return value every time it is called.
# Use *args, **kwargs and functools.wraps. Apply it to two functions with different signatures to prove it is generic.
```

```python
from functools import wraps
def log_calls(fun):

    @wraps(fun)
    def wrapper(*args,**kwargs):
        result = fun(*args,**kwargs)
        print(result)
        return result
    return wrapper
        


@log_calls
def add(a,b):
    return a+b
add(5,9)
```
