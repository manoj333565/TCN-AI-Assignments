```python
#Q1.(1) survival rate by embark_town
df.groupby("embark_town")["survived"]
# print(df)

survived_rate = df.groupby("embark_town")["survived"].mean()
print(survived_rate)
```

```python
# (2) median fare by class and alone

median_fare=df.groupby(['class','alone'])["fare"].median()
median_fare
```
